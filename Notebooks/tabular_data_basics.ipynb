```python
# Dataset summary
print("Total Bookings:", df["Bookings"].sum())

print("Total Refunds:", df["Refunds"].sum())

print("Highest Booking Category:")
print(df.loc[df["Bookings"].idxmax()])

print("Categories with more than 100 bookings:")
print(df[df["Bookings"] > 100])
```

PDF

```python
import pdfplumber
import pandas as pd

pdf_path = "../Data/Uploads/bookings.pdf"

with pdfplumber.open(pdf_path) as pdf:
    table = pdf.pages[0].extract_table()

pdf_df = pd.DataFrame(table[1:], columns=table[0])

pdf_df["Bookings"] = pd.to_numeric(pdf_df["Bookings"])
pdf_df["Refunds"] = pd.to_numeric(pdf_df["Refunds"])

print(pdf_df)
```

```python
def analyze_bookings(df):

    total_bookings = df["Bookings"].sum()
    total_refunds = df["Refunds"].sum()

    highest_category = df.loc[df["Bookings"].idxmax()]

    return {
        "total_bookings": int(total_bookings),
        "total_refunds": int(total_refunds),
        "highest_booking_category": highest_category["Category"],
        "highest_bookings": int(highest_category["Bookings"])
    }


print("CSV Analysis:")
print(analyze_bookings(df))

print("\nExcel Analysis:")
print(analyze_bookings(excel_df))

print("\nPDF Analysis:")
print(analyze_bookings(pdf_df))
```
